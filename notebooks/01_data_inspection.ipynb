```python
# Import libraries
import pandas as pd
```

```python
# Import data
online_retail = pd.read_excel(r"D:\Portfolio\retail-online-analysis\data\Online Retail.xlsx")
```

```python
# 1. Data Dimensions
print("Shape:", online_retail.shape)
```

```python
# 2. The names of the eight columns in the dataset
print("Column Names:", online_retail.columns.tolist())
```

```python
# 3. Data Types of the eight columns in the dataset
print("Data Types:\n", online_retail.dtypes)
```

```python
# 4. The number of unique values in each of the eight columns in the dataset
columns = [
    'InvoiceNo',
    'StockCode',
    'Description',
    'CustomerID',
    'Country'
]

print("Unique Values:\n", online_retail[columns].nunique())
```

```python
# 5. Missing Values across the eight columns in the dataset
missing_values = online_retail.isna().sum()
print("Missing Values:\n", missing_values)
```

```python
# 6. First 5 Rows
print("First 5 Rows:\n", online_retail.head())
```

```python
# 7. Duplicate rows
duplicate_rows = online_retail.duplicated().sum()
print("Number of duplicate rows:", duplicate_rows)
```

```python
# 8. Negative quantities
negative_quantities = online_retail[online_retail['Quantity'] < 0]
print("Negative Quantities:", negative_quantities.shape[0])
```

```python
# 9. Non-positive UnitPrice
non_positive_unit_price = online_retail[online_retail['UnitPrice'] <= 0]
print("Non-positive UnitPrice:", non_positive_unit_price.shape[0])
```

```python
# 10. Cancellations
cancellations = online_retail[online_retail['InvoiceNo'].str.startswith('C', na=False)]
print("Cancellations:", cancellations.shape[0])
```

```python
# 11. Rows that are both Cancellations and Negative Quantities
cancellations_and_negatives = online_retail[
    (online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

print("Cancellations and Negative Quantities:", cancellations_and_negatives.shape[0])
```

```python
# Print the cancellations and negative quantities DataFrame
print(cancellations_and_negatives.head())
```

```python
# 12. Rows that are both Non-Cancellations and Negative Quantities
non_cancellations_and_negative_quantities = online_retail[
    (~online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

print("Non-Cancellations and Negative Quantities:", non_cancellations_and_negative_quantities.shape[0])
```

```python
# Print the non-cancellations and negative quantities DataFrame
print(non_cancellations_and_negative_quantities.head())
```

```python
print(non_cancellations_and_negative_quantities.tail())
```

```python
# 13. Rows that are both Non-Cancellations and Negative Quantities with no NaN in its Description
non_cancellations_and_negatives = online_retail[
    (~online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

notnan_non_cancellations_and_negatives = non_cancellations_and_negatives.dropna(subset=['Description'])
print("Not NaN Non-Cancellations and Negative Quantities:", notnan_non_cancellations_and_negatives.shape[0])
```

```python
print(notnan_non_cancellations_and_negatives.head())
```

```python
print(notnan_non_cancellations_and_negatives.tail())
```

```python
# 14. Rows that are both Non-Cancellations and Negative Quantities with NaN in its Description
non_cancellations_and_negatives = online_retail[
    (~online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

nan_non_cancellations_and_negatives = non_cancellations_and_negatives[
    non_cancellations_and_negatives["Description"].isna()
]

print("NaN Non-Cancellations and Negative Quantities:", nan_non_cancellations_and_negatives.shape[0])
```

```python
print(nan_non_cancellations_and_negatives.head())
```

```python
# 15. Rows that are both Cancellations and Negative Quantities with NaN in its Description
cancellations_and_negative_quantities = online_retail[
    (online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

nan_cancellations_and_negatives = cancellations_and_negative_quantities[cancellations_and_negative_quantities["Description"].isna()]

print("NaN Cancellations and Negative Quantities:", nan_cancellations_and_negatives.shape[0])
```

```python
# 16. Remaining NaN Descriptions after excluding NaN Descriptions in Non-Cancellations and Negative Quantities

nan_descriptions = online_retail[
    online_retail["Description"].isna()
]

remaining_nan_descriptions = nan_descriptions[
    ~nan_descriptions.index.isin(
        nan_non_cancellations_and_negatives.index
    )
]

print("Remaining NaN Descriptions:", remaining_nan_descriptions.shape[0])
```

```python
# 17. Rows that are both Cancellations and Non-positive Unit Prices
cancellations_and_negative_unit_prices = online_retail[
    (online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['UnitPrice'] <= 0)
]

print("Cancellations and Non-positive UnitPrice:", cancellations_and_negative_unit_prices.shape[0])
```

```python
# 18. Rows that are both Non-Cancellations and Non-positive Unit Prices
non_cancellations_and_negative_unit_prices = online_retail[
    ~(online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['UnitPrice'] <= 0)
]

print("Non-Cancellations and Non-positive UnitPrice:", non_cancellations_and_negative_unit_prices.shape[0])
```

```python
# 19. Rows that are both Non-Cancellations and Non-positive Unit Prices and Quantities
non_cancellations_and_negative_unit_prices_and_quantities = online_retail[
    ~(online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['UnitPrice'] <= 0) & (online_retail['Quantity'] < 0)
]

print("Non-Cancellations and Non-positive UnitPrice and quantities:", non_cancellations_and_negative_unit_prices_and_quantities.shape[0])
```

```python
# 20. Rows that are both Non-Cancellations and Non-positive Unit Prices and Non-negative Quantities

remaining_unit_price_records = online_retail[
    ~(online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['UnitPrice'] <= 0) & (online_retail['Quantity'] > 0)
]

print("Remaining records:", remaining_unit_price_records.shape[0])
```

```python
# 21. Rows that are Zero Quantities
zero_quantities = online_retail[
    online_retail['Quantity'] == 0
]

print("Zero Quantities:", zero_quantities.shape[0])
```

```python
# 22. Date range
online_retail['InvoiceDate'].min()
online_retail['InvoiceDate'].max()

print("Date Range: ", online_retail['InvoiceDate'].min(), " to ", online_retail['InvoiceDate'].max())
```
