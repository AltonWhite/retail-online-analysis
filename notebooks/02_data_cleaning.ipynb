```python
# Import libraries
import pandas as pd
```

```python
# Import data
online_retail = pd.read_excel(r"D:\Portfolio\retail-online-analysis\data\Online Retail.xlsx")
```

```python
# 1. Cancellations (Reprise)
cancellations = online_retail[online_retail['InvoiceNo'].str.startswith('C', na=False)]
print("Cancellations:", cancellations.shape[0])
```

```python
# 2. Negative Quantities (Reprise)
negative_quantities = online_retail[online_retail['Quantity'] < 0]
print("Negative Quantities:", negative_quantities.shape[0])
```

```python
# 3. Cancellations and Negative Quantities (Reprise)
cancellations_and_negative_quantities = online_retail[
    (online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

print("Cancellations and Negative Quantities:", cancellations_and_negative_quantities.shape[0])
```

```python
# 3. Cancellations and Positive Quantities (Reprise)
cancellations_and_positive_quantities = online_retail[
    (online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] > 0)
]

print("Cancellations and Positive Quantities:", cancellations_and_positive_quantities.shape[0])
```

```python
# 4. Non-Cancellations and Negative Quantities (Reprise)
non_cancellations_and_negative_quantities = online_retail[
    (~online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] < 0)
]

print("Non-Cancellations and Negative Quantities:", non_cancellations_and_negative_quantities.shape[0])
```

```python
# 5. Non-Cancellations and Negative Quantities Unit Price Distribution

non_cancellation_and_negative_qualities_unitprices = non_cancellations_and_negative_quantities['UnitPrice'].value_counts()

print("Non-Cancellations and Negative Quantities Unit Prices:" , non_cancellation_and_negative_qualities_unitprices)
```

```python
# 6. Non-Cancellations and Negative Quantities Descriptions

non_cancellation_and_negative_qualities_descriptions = non_cancellations_and_negative_quantities['Description'].value_counts(dropna=False)

print("Non-Cancellations and Negative Quantities Unit Price Distribution:" , non_cancellation_and_negative_qualities_descriptions)
```

```python
# 7. Non-cancellations, positive_quantities, and positive unitprice
non_cancellations_positive_quantities_and_positive_unitprice = online_retail[
    (~online_retail['InvoiceNo'].str.startswith('C', na=False)) & (online_retail['Quantity'] > 0) & (online_retail['UnitPrice'] > 0)
]
print("Non-cancellations, positive quantities, and positive unitprice:", non_cancellations_positive_quantities_and_positive_unitprice.shape[0])
```

```python
# Copy online retail to sales_data before the real cleaning begins
sales_data = online_retail.copy()
```

```python
# 1. Filter non-cancellations, positive quantities, and positive unitprice

sales_data = sales_data[
    (~sales_data['InvoiceNo'].str.startswith('C', na=False)) & (sales_data['Quantity'] > 0) & (sales_data['UnitPrice'] > 0)
]

print("Original rows:", online_retail.shape[0])
print("Sales data rows:", sales_data.shape[0])
```

```python
# 2. Check for cancellations, non-positive quantities, and non-positive unit prices in the filtered sales_data
print("Cancellations:", sales_data['InvoiceNo'].str.startswith('C', na=False).sum())
print("Non-positive quantities:", (sales_data['Quantity'] <= 0).sum())
print("Non-positive unit prices:", (sales_data['UnitPrice'] <= 0).sum())
```

```python
# 3. Check for duplicates in sales_data

duplicate_rows = sales_data.duplicated().sum()
print("Number of duplicate rows:", duplicate_rows)
```

```python
# 4. Deduplicate sales_data
sales_data_deduped = sales_data.drop_duplicates()

print("Deduped row count:", sales_data_deduped.shape[0])
print("Number of duplicate rows:", sales_data_deduped.duplicated().sum())
```

```python
# 5. Check for NaN values in sales_data_deduped
print("Number of NaN rows:\n", sales_data_deduped.isna().sum())
```

```python
# Print the first few rows
print(sales_data_deduped.head())
```

```python
# Print the last few rows
print(sales_data_deduped.tail())
```

```python
# The number of unique values in each of the eight columns in the dataset
columns = [
    'InvoiceNo',
    'StockCode',
    'Description',
    'CustomerID',
    'Country'
]

print("Unique Values:\n", sales_data_deduped[columns].nunique())
```

```python
# 6. Save sales_data_deduped to CSV
sales_data_deduped.to_csv(r"D:\Portfolio\retail-online-analysis\data\sales_data_deduped.csv", index=False)
```

# Summary

The data-cleaning process consisted of the following steps:
- **Transaction filtering**: Cancellations, negative quantities, and non-positive unit prices were filtered. There were no zero-quantity records. The dataset was reduced from 541,909 to 530,104 rows.
- **Duplicate removal**: Exact duplicate rows were dropped, reducing the dataset further to 524,878 rows.
- **Data export**: The filtered and deduplicated data was saved as sales_data_deduped.csv for further analysis in the other notebooks.
