```python
import pandas as pd

sales_data_deduped = pd.read_csv(
    r"D:\Portfolio\retail-online-analysis\data\sales_data_deduped.csv",
    dtype={'InvoiceNo': 'string'}
)

sales_data_deduped['InvoiceDate'] = pd.to_datetime(sales_data_deduped['InvoiceDate'])
sales_data_deduped['CustomerID'] = sales_data_deduped['CustomerID'].astype('Int64')

sales_data_deduped['Revenue'] = sales_data_deduped['Quantity'] * sales_data_deduped['UnitPrice']
```

```python
sales_data_deduped['Description'].unique()
```

```python
# Create a DataFrame that excludes rows with non-merchandise descriptions in the 'Description' column
non_product_items = [
    'Adjust bad debt',
    'AMAZON FEE',
    'DOTCOM POSTAGE',
    'POSTAGE',
    'Manual',
    'Voucher',
    'SAMPLE'
]

product_sales = sales_data_deduped[
    ~sales_data_deduped['Description'].str.contains(
        '|'.join(non_product_items), case=False, na=False
    )
].copy()
```

```python
print(product_sales.head())
```

```python
print(product_sales.tail())
```

```python
# Count rows before and after filtering
rows_before = sales_data_deduped.shape[0]
rows_after = product_sales.shape[0]

print(f"Rows before filtering: {rows_before}")
print(f"Rows after filtering: {rows_after}")
```

```python
# The number of unique values in each of the eight columns in the dataset
columns = [
    'InvoiceNo',
    'StockCode',
    'Description',
    'CustomerID',
    'Country'
]

print("Unique Values:\n", product_sales[columns].nunique())
```

```python
# Export to a new CSV file
product_sales.to_csv(r"D:\Portfolio\retail-online-analysis\data\product_sales.csv", index=False)
```

# Summary

The product descriptions were examined to identify records that did not represent physical merchandise, such as postage charges, amazon fees, manual, samples, and gift vouchers. These non-merchandise records were excluded to create a separate merchandise-only dataset, which was saved as product_sales.csv. As a result, the dataset was reduced from 524,878 to 522,692 rows.
